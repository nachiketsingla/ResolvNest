#Question-1

```python
roll_number_str="1024170077"
#L=[int(digit) * 10 for digit in roll_number_str]
L=[]
for digit in roll_number_str:
  number=int(digit)
  L.append(number*10)

#i.print L
print("Initial List L:",L)
```

```python
#ii. Add two numbers (append & insert)
L.append(99)
L.insert(2,88)
print("L after append and insert",L)
```

```python
#iii. Remove two elements (remove & pop)
L.remove(20)
print("L after remove:",L)
popped_element=L.pop(3)
print("L after pop(3) :",L)
```

```python
#iv. sort L
L.sort()
print('L sorting ascending',L)
L.sort(reverse=True)
print("L sorting descending",L)
```

```python
# v. Slicing first 3 and last 3
print("1.v. First three elements:", L[:3], "| Last three elements:", L[-3:])
```

```python
#vi. List comprehension for elemenent> average
avg_L=sum(L)/len(L)
print(L)
print('average=',avg_L)
L_greater=[x for x in L if x> avg_L]
print(L_greater)
```

```python
initial_8=[int(digit)*10 for digit in roll_number_str[:8]]
scores=tuple(initial_8)
print("scores tuple:",scores)
```

```python
#i. Highest and lowest score analytics
highest_score=max(scores)
highest_indx=scores.index(highest_score)
lowest_score=min(scores)
lowest_count=scores.count(lowest_score)
print(f" Highest :{highest_score} at index{highest_indx}. Lowest: {lowest_score} (appears {lowest_count} times)")
```

```python
#ii. reverse tuple to list
reversed_scores_list=list(scores)[::-1]
print("reversed tuple as list",reversed_scores_list)
```

```python
#iii. user input search
user_score=int(input("enter a score to reach: "))
if user_score in scores:
  print(f' score {user_score} found at index:',scores.index(user_score))
else:
  print(f' score {user_score} is not present')
```
